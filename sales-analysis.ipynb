```python
import pandas as pd

# Load your CSV data
sales_data = pd.read_csv('/kaggle/input/sales-forecasting/train.csv')

# Convert the 'Order Date' column to a datetime format
sales_data['Order Date'] = pd.to_datetime(sales_data['Order Date'], format='%d/%m/%Y')

#Extract the year from the date column
sales_data['Year'] = sales_data['Order Date'].dt.year

# Group by year and calculate total sales revenue
yearly_sales_revenue = sales_data.groupby('Year')['Sales'].sum().reset_index()

# Sort the result in descending order of sales
yearly_sales_revenue = yearly_sales_revenue.sort_values('Sales', ascending=False)

# Display the result
print(yearly_sales_revenue)
```

```python
print("Available columns:", sales_data.columns.tolist())
print(sales_data[['Year', 'Product Name', 'Sales']].head()) # Show first few rows 
```

```python
# Group by Year and Product Name to calculate total sales
yearly_product_sales = sales_data.groupby(['Year', 'Product Name'])['Sales'].sum().reset_index()

# Sort by year and sales in descending order
top_products_by_year = yearly_product_sales.sort_values(['Year', 'Sales'], ascending=[True, False])

# Display the result
print(top_products_by_year)
```

```python
# Extract Quarter from 'Order Date'
sales_data['Quarter'] = sales_data['Order Date'].dt.quarter

# Group by Year and Quarter to calculate total sales
quarterly_sales = sales_data.groupby(['Year', 'Quarter'])['Sales'].sum().reset_index()

# Sort by total sales in descending order to see the most successful quarters
quarterly_sales = quarterly_sales.sort_values(['Sales'], ascending=False)

# Display the result
print(quarterly_sales)
```

```python
import matplotlib.pyplot as plt

# Create a line plot of quarterly sales
plt.figure(figsize=(10,6))
for year in quarterly_sales['Year'].unique():
    subset = quarterly_sales[quarterly_sales['Year'] == year]
    plt.plot(subset['Quarter'], subset['Sales'], marker='o', label=f"Year{year}")

plt.title("Quarterly Sales Trends Over the Years")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")
plt.xticks([1, 2, 3, 4]) # Ensure quarters are labeled correctly
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Create a bar chart of the most successful quarters
plt.figure(figsize=(10, 6))
plt.bar(quarterly_sales['Quarter'].astype(str) + " (Year " + quarterly_sales['Year'].astype(str) + ")",
       quarterly_sales['Sales'], color='skyblue')

plt.title("Most Successful Quarters by Total Sales")
plt.xlabel("Quarter (Year)")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability 
plt.show()
```

```python
import pandas as pd

# Extract the Month from 'Order Date'
sales_data['Month'] = sales_data['Order Date'].dt.month

# Group by Year and Month to calculate the sales
monthly_sales = sales_data.groupby(['Year', 'Month'])['Sales'].sum().reset_index()

# Sort the data by Year and Month for beter visualization
monthly_sales = monthly_sales.sort_values(['Year', 'Month'])

# Display the result
print(monthly_sales.head(12)) # Show first 12 months as a preview
```

```python
import matplotlib.pyplot as plt

# Create a line plot of Monthly sales
plt.figure(figsize=(10,6))
for year in monthly_sales['Year'].unique():
    subset = monthly_sales[monthly_sales['Year'] == year]
    plt.plot(subset['Month'], subset['Sales'], marker='o', label=f"Year{year}")

plt.title("Monthly Sales Trends Over the Years")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(range(1, 13), ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']) # Label months
plt.legend()
plt.grid(True)

# Save the figure as an image
plt.savefig ("monthly_sales_trends.png", dpi=300)

plt.show()
```

```python
monthly_sales.to_csv("monthly_sales.csv", index=False)
```

```python
import pandas as pd

# Group by Year to calculate total sales
yearly_sales_growth = sales_data.groupby('Year')['Sales'].sum().reset_index()

# Calculate Year-over-Year Growth 
yearly_sales_growth['YoY Growth (%)'] = yearly_sales_growth['Sales'].pct_change() * 100

# Sort by year in ascending order
yearly_sales_growth = yearly_sales_growth.sort_values('Year')

# Display the results
print(yearly_sales_growth)
```

```python
import matplotlib.pyplot as plt


# Create a bar chart of the yearly sales
plt.figure(figsize=(10, 6))
plt.bar(yearly_sales_growth['Year'], yearly_sales_growth['Sales'], color='skyblue')

plt.title("Total Sales by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(yearly_sales_growth['Year'])

# Save the figure as a PNG file
plt.savefig("yearly_sales_growth.png", dpi=300)

plt.show()
```

```python
yearly_sales_growth.to_csv("yearly_sales_growth.csv", index=False)
```
